```python
import pandas as pd

df = pd.read_csv(
    "/content/IMDB_dataset.csv",
    engine="python",
    on_bad_lines="skip"
)

print(df.shape)
df.head()
```

```python
print(df.shape)
print(df.columns)
df.head()
```

```python
print(df.isnull().sum())
```

```python
def clean_text(text):
    text = text.lower()
    return text

df["cleaned_text"] = df["review"].apply(clean_text)
```

```python
print(df["review"].iloc[0])
print(df["cleaned_text"].iloc[0])
```

```python
def expand_contractions(text):
    return contractions.fix(text)

df["expanded_text"] = df["cleaned_text"].apply(expand_contractions)

print(df["expanded_text"].iloc[0])
```

```python
def remove_unwanted(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text

df["cleaned_text"] = df["expanded_text"].apply(remove_unwanted)

print(df["cleaned_text"].iloc[0])
```
